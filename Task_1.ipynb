This conclusion was drawn by analyzing the **common symptoms observed in patients diagnosed at stage 2**, which showed similar patterns across this group.

- **Cancer generally starts at stage 1** in all patients, but if not detected early, it progresses to more severe stages. Even patients living in relatively **healthy environments** were diagnosed at stage 2 or later, indicating that **delayed detection**—not just environmental factors—played a significant role.

- Further analysis of patients diagnosed at **stage 2**, despite having **low smoking levels**, revealed two categories:
  - One group with relatively **healthy environments and lifestyles**
  - Another group from relatively **worse conditions**

- **Surprisingly, both groups displayed nearly identical symptoms**, with only **minute variations in symptom intensity**—differing by just **1 unit**. This suggests that the **severity and progression of cancer symptoms** were largely **independent of their living conditions** and were more likely the result of **delayed diagnosis**.

```python
grouped = df.groupby("Level")[["OccuPational Hazards","Air Pollution","Passive Smoker"]].agg("mean")
grouped.reset_index(inplace = True)
melted = grouped.melt(id_vars="Level", var_name="Exposure Type", value_name="Mean Value")
```

```python
sns.barplot(
    data=melted,
    x="Level",
    y="Mean Value",
    hue="Exposure Type",
)
```

* patients who got smoking or alcohol level above 8 got level 3 or level 2 cancer
* the variation in the intensity of cancer is due to the passive smoking ,occupatioal hazards , genetic risk, and air pollution, we can generally say it is due to the enviornment they are living in
