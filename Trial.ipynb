```python
import pandas as pd

data = {
    'Student': ['Arjun', 'Bhavana', 'Charan'],
    'Maths': [88, 92, 95],
    'Science': [84, 89, 90],
    'English': [90, 85, 88]
}

df = pd.DataFrame(data)
print(df)
```

```python
df_melt = pd.melt(df,id_vars = "Student",value_vars = ["Maths","Science","English"],var_name="Subject",value_name="Marks")
df_melt
```

```python
df_melt.groupby("Subject")["Marks"].mean()
```
